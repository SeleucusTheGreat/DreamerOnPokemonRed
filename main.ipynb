### Testing Loop

```python

import torch
from PokemonRedEnv import create_envs
from policy import Policy

RUN_MODE = "train" #train or eval
ROM_PATH = "PokemonRed.gb"
NUM_EVAL_EPISODES = 1
NUM_ENVS = 5
         # environments used for training
NUM_EVAL_ENVS = 4    # environments opened in eval mode (rendered with SDL2)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# In eval mode: show the game (SDL2 window) at real-time speed (1) with NUM_EVAL_ENVS envs.
# In train mode: headless (null) at unlimited speed (0) with NUM_ENVS envs.
if RUN_MODE == "eval":
    num_envs, window, speed = NUM_EVAL_ENVS, "SDL2", 1
else:
    num_envs, window, speed = NUM_ENVS, "null", 0

envs = create_envs(num_envs=num_envs, rom_path=ROM_PATH, state_dir="StartingFiles", wind=window, speed=speed)
print("Initializing Dreamer Policy...")

policy = Policy(envs=envs, device=device, visualize_dreams=True)

if RUN_MODE == "train":
    print("Starting Training Mode...")
    policy.train()

elif RUN_MODE == "eval":
    print("Starting Evaluation Mode...")
    policy.evaluate(num_episodes=NUM_EVAL_EPISODES)
else:
    print(f"Invalid RUN_MODE: '{RUN_MODE}'. Please set it to 'train' or 'eval'.") 

print("All environments safely closed.")
```
